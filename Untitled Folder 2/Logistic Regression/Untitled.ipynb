The objective of logistic regression in this study is to model the probability of students successfully passing the exam based on demographic, socioeconomic, and academic performance features.

 For example, we have input for the forecast probability of passing/not passing.
X:
Category features which converted a number
gender
race/ethnicity
parental level of education	
lunch
test preparation course
Numeric features
math score
reading score
writing score

First we will build a regression based on the input data X
$$z=X^T*w$$
$$z=w_1*x_1+w_2*x_2+...+w_n*x_n + b$$

We introduce the sigmoid function into the obtained Z value
$$y_{pred}=a=\sigma(z)$$
$$\sigma(z) = \frac{1}{1+e^{-z}}$$

Now we calculate the loss function:
$$L=-ylog(a) + (1-y)log(1-a)$$

To update the weights w,b we implement backpropagation through partial derivatives:
$$\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a}*\frac{\partial a}{\partial z}*\frac{\partial z}{\partial w_1}$$
$$\frac{\partial L}{\partial a} = \frac{-y}{a} + \frac{1-y}{1-a}$$
$$\frac{\partial a}{\partial z} = a(1-a)$$
$$\frac{\partial z}{\partial w_1} = x_1$$
$$\frac{\partial L}{\partial w_1} = ((\frac{-y}{a} + \frac{1-y}{1-a}) * a(1-a))*x_1 = (a-y)*w_1$$ 

$$w_1 = w_1 - \alpha*\frac{\partial L}{\partial w_1}$$
$$b = b- \alpha*\frac{\partial L}{\partial b}$$

Now we need to classify by probabilities
if 
$$p(z)>=0.5$$
Pass:
if
$$p(z)<0.5$$
not pass

```python
columns = ['gender', 'race/ethnicity', 'parental level of education', 'lunch', 'test preparation course']

for col in columns:
    labels = {val: i for i, val in enumerate(dataset[col].unique())}
    dataset[col] = dataset[col].replace(labels)
    print(f"{col} үшін ауыстыру кестесі: {labels}")

print(dataset.head())
```

```python
cols = ['math score', 'reading score', 'writing score']

for col in cols:
    col_mean = dataset[col].mean()
    col_std = dataset[col].std()
    
    # Формула: (x - mean) / std
    dataset[col] = (dataset[col] - col_mean) / col_std

print(dataset[cols].head())
```

```python
# X input gender race/ethnicity parental level of education lunch test preparation course math score reading score writing score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
X = dataset.values
y = np.random.randint(0, 2, size=(1000, 1))

def weightInitialization(n_features):   
    # Используем маленькие случайные значения
    w = np.random.normal(loc=0, scale=0.01, size=(n_features, 1)) 
    b = np.zeros((1,)) # Смещение (bias) обычно инициализируют нулем
    return w, b
def sigmoid_activation(result):
    final_result = 1/(1+np.exp(-result))
    return final_result

def model_optimize(w, b, X, y):
    m = X.shape[0]

    # 1. Prediction: X (1000, 8) * w (8, 1) + b -> результат (1000, 1)
    # Это классическая формула: z = Xw + b
    final_result = sigmoid_activation(np.dot(X, w) + b)
    
    # 2. Cost (Log Loss)
    # Используем маленькое число 1e-15, чтобы не было log(0) = -inf
    cost = (-1/m) * np.sum(y * np.log(final_result + 1e-15) + (1 - y) * np.log(1 - final_result + 1e-15))

    # 3. Gradients
    dw = (1/m) * np.dot(X.T, (final_result - y)) # Результат (8, 1)
    db = (1/m) * np.sum(final_result - y)

    grads = {"dw": dw, "db": db}
    return grads, cost

def model_predict(w, b, X, y, learning_rate=0.01, no_iterations=2000):
    costs = []
    for i in range(no_iterations):
        grads, cost = model_optimize(w, b, X, y)
        
        # Обновляем параметры: веса и градиенты теперь одной формы (8, 1)
        w = w - (learning_rate * grads["dw"])
        b = b - (learning_rate * grads["db"])

        if i % 100 == 0:
            costs.append(cost)

    return {"w": w, "b": b}, costs
    #final parameters
    coeff = {"w": w, "b": b}
    gradient = {"dw": dw, "db": db}

    return coeff, gradient, costs

```
