The objective of logistic regression in this study is to model the probability of students successfully passing the exam based on demographic, socioeconomic, and academic performance features.

 For example, we have input for the forecast probability of passing/not passing.
X:
Category features which converted a number
gender
race/ethnicity
parental level of education	
lunch
test preparation course
Numeric features
math score
reading score
writing score

First we will build a regression based on the input data X
$$z=X^T*w$$
$$z=w_1*x_1+w_2*x_2+...+w_n*x_n + b$$

We introduce the sigmoid function into the obtained Z value
$$y_{pred}=a=\sigma(z)$$
$$\sigma(z) = \frac{1}{1+e^{-z}}$$

Now we calculate the loss function:
$$L=-ylog(a) + (1-y)log(1-a)$$

To update the weights w,b we implement backpropagation through partial derivatives:
$$\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a}*\frac{\partial a}{\partial z}*\frac{\partial z}{\partial w_1}$$
$$\frac{\partial L}{\partial a} = \frac{-y}{a} + \frac{1-y}{1-a}$$
$$\frac{\partial a}{\partial z} = a(1-a)$$
$$\frac{\partial z}{\partial w_1} = x_1$$
$$\frac{\partial L}{\partial w_1} = ((\frac{-y}{a} + \frac{1-y}{1-a}) * a(1-a))*x_1 = (a-y)*w_1$$ 

$$w_1 = w_1 - \alpha*\frac{\partial L}{\partial w_1}$$
$$b = b- \alpha*\frac{\partial L}{\partial b}$$

Now we need to classify by probabilities
if 
$$p(z)>=0.5$$
Pass:
if
$$p(z)<0.5$$
not pass
