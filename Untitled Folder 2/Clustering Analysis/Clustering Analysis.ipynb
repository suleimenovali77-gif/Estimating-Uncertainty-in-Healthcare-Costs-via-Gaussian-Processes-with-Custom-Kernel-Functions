The goal of this project is to divide students into 3 clusters: those with high grades in all subjects, those with average grades, and those with low grades.

K-means clustering. So, what is it? The essence of K-means (the k-means method) in this context is the transition from a "heap of data" to a mathematically sound typology of students.
Objective
1. Transition from a continuum to discrete groups
Student grades are distributed continuously. K-means takes this point cloud and finds "centers of attraction" (centroids) within it.

Meaning: We stop looking at each student individually and begin to see the structure of the population. Instead of 1,000 different people, we get three clear profiles: "Strong," "Average," and "Weak."

2. Minimization of variability (Compactness)
Mathematically, K-means aims to make students within a cluster as similar to each other as possible, and the clusters themselves as distant from each other as possible.

Cost Function (Inertia): We minimize the sum of the squared distances from students to the center of their group. The lower this value, the more homogeneous the composition of the top or bottom students.

3. Automatically detecting boundaries
In a typical analysis, we would say, "Those with an average score > 80 are top students." But that's subjective.

K-means does this objectively: It automatically draws boundaries (Voronoi diagrams) where the data density changes. It finds a "natural" threshold between groups based on the actual distribution of scores in your StudentsPerformance.csv, not on our expectations.

We also use the PCA tool, which is the principal component method, which compresses our data and leaves only the essence, which is useful in clustering.

Objective

1. Combating the "Curse of Dimensionality" In a high-dimensional space (if we had 50 parameters instead of 3), the distance between any two points tends to a constant. Problem: The Euclidean distance (the basis of K-means) no longer distinguishes between "close" and "distant" data. PCA Solution: We retain only those dimensions where the data are truly "scattered" (maximum variance), pruning those where they hardly change. This makes clusters more pronounced. 2. Removing Multicollinearity In your dataset, math, reading, and writing scores are highly correlated. From a geometric perspective, your data is stretched along a single narrow diagonal. Problem: K-means treats each feature as an independent vector. If features are correlated, the same information (say, "general intelligence") is counted three times, giving it an unreasonably large "weight" when calculating the distance. PCA Solution: It creates decorrelated (orthogonal) components. Each new axis in PCA is independent of the others. 3. Noise Filtering: Data variance can be decomposed into: Signal: Main patterns (e.g., overall academic performance). Noise: Random fluctuations (e.g., a student did poorly on a writing test). PCA sorts components by their eigenvalues ​​$\lambda$. Small $\lambda$ values ​​usually correspond to noise. By discarding them, we force K-means to cluster only by the "essence" of the data. 4. Mathematical Approximation (Ding & He Relationship)There is a proven theorem that the subspace generated by the first $k-1$ principal components is a continuous solution to the K-means problem. In simple terms, PCA finds the "plane" on which the clusters will be most separated from each other even before we run the K-means iterations.

Mathematical Justification: PCA + K-means Clustering1. Data Preparation: StandardizationBefore applying PCA, it is necessary to standardize the features. For a data matrix $X \in \mathbb{R}^{n \times d}$:$$z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}$$This ensures that each feature (Math, Reading, Writing) has $\mu=0$ and $\sigma^2=1$, preventing features with high variance from dominating.2. Dimensionality Reduction (PCA)Principal component analysis finds an orthogonal basis that maximizes the explained variance.Step A: Covariance MatrixCalculate the $\Sigma$ matrix:$$\Sigma = \frac{1}{n-1} Z^T Z$$Step B: Spectral DecompositionSolve the eigenvalue problem:$$\Sigma v_k = \lambda_k v_k$$Where $\lambda_k$ is the variance along the $k$-th component. We select $m$ vectors corresponding to the largest $\lambda$. Step B: Projection The new feature space $Z_{pca} \in \mathbb{R}^{n \times m}$:$$Z_{pca} = Z \cdot W_m, \quad W_m = [v_1, v_2, \dots, v_m]$$3. K-means clustering We minimize the WCSS (Within-Cluster Sum of Squares) functional on the obtained components. Objective function For a set of clusters $S = \{S_1, S_2, \dots, S_K\}$ we seek:$$\arg \min_{S} \sum_{i=1}^{K} \sum_{z \in S_i} \| z - \mu_i \|^2$$Where $\mu_i$ is the centroid of the cluster $S_i$:$$\mu_i = \frac{1}{|S_i|} \sum_{z \in S_i} z$$

Also we can calculate Silhouette Coefficient  . Mathematical definition of Silhouette CoefficientFor each point $i$, two quantities are calculated:$a(i)$ (Cohesion): The average distance from point $i$ to all other points in the same cluster.$$a(i) = \frac{1}{|C_A| - 1} \sum_{j \in C_A, i \neq j} d(i, j)$$The smaller $a(i)$, the better the point is "fitted" to its neighbors.$b(i)$ (Separation): Average distance from point $i$ to points of the nearest (closest of the aliens) cluster $C_B$.$$b(i) = \min_{C_k \neq C_A} \frac{1}{|C_k|} \sum_{j \in C_k} d(i, j)$$The larger $b(i)$, the further the point is from its neighbors. Silhouette coefficient $s(i)$:$$s(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}}$$

```python
#2
cols = ['math score', 'reading score', 'writing score']

for col in cols:
    col_mean = dataset[col].sum() / len(dataset)
    col_std = np.sqrt(np.sum((dataset[col] - col_mean)**2) / len(dataset))

    dataset[col] = (dataset[col] - col_mean) / col_std
    
    # Формула: (x - mean) / std
print(dataset[cols].head())
```

```python
dataset
```

```python

X = dataset[cols].values


mean_vector = np.mean(X, axis=0)

X_centered = X - mean_vector
n = X_centered.shape[0]


cov_matrix = (X_centered.T @ X_centered) / (n - 1)

print("Матрица ковариации:\n", cov_matrix)

eigen_values, eigen_vectors = np.linalg.eig(cov_matrix)


idx = eigen_values.argsort()[::-1]
eigen_values = eigen_values[idx]
eigen_vectors = eigen_vectors[:, idx]


W = eigen_vectors[:, :2]


X_pca = X.dot(W)

print("Объясненная дисперсия (в %):", (eigen_values / np.sum(eigen_values)) * 100)
```

```python

dataset['PC1'] = X_pca[:, 0]
dataset['PC2'] = X_pca[:, 1]

dataset[['math score', 'reading score', 'writing score', 'PC1', 'PC2']].head()
```

```python
# Инициализация: берем 3 случайные точки из PCA-данных
k = 3
centroids = X_pca[np.random.choice(X_pca.shape[0], k, replace=False)]

for i in range(100): 
    distances = np.sqrt(((X_pca[:, np.newaxis, :] - centroids)**2).sum(axis=2))
    
    labels = np.argmin(distances, axis=1)
    
    
    new_centroids = np.array([X_pca[labels == j].mean(axis=0) for j in range(k)])
    
    if np.allclose(centroids, new_centroids):
        print(f"Конвергенция достигнута на шаге {i}")
        break
    centroids = new_centroids
```

```python
dataset['Cluster'] = labels
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

# Рисуем точки, раскрашивая их в зависимости от кластера
scatter = plt.scatter(dataset['PC1'], dataset['PC2'], 
                      c=dataset['Cluster'], cmap='viridis', s=50, alpha=0.7)

# Добавляем подписи
plt.title('Результат кластеризации в пространстве PCA', fontsize=15)
plt.xlabel(f'PC1 (Объясняет {eigen_values[0]/sum(eigen_values)*100:.1f}% дисперсии)')
plt.ylabel(f'PC2 (Объясняет {eigen_values[1]/sum(eigen_values)*100:.1f}% дисперсии)')
plt.colorbar(scatter, label='Кластер')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
def calculate_silhouette(X, labels):
    n = len(X)
    s_scores = np.zeros(n)
    
    # Расстояние между всеми парами точек (Евклид)
    # Для больших данных лучше использовать sklearn, но для понимания:
    from scipy.spatial.distance import cdist
    dist_matrix = cdist(X, X, metric='euclidean')
    
    for i in range(n):
        curr_cluster = labels[i]
        
        # 1. Считаем a(i)
        same_cluster_idx = np.where(labels == curr_cluster)[0]
        if len(same_cluster_idx) > 1:
            a_i = np.mean(dist_matrix[i, same_cluster_idx[same_cluster_idx != i]])
        else:
            a_i = 0
            
        # 2. Считаем b(i)
        other_clusters = np.unique(labels[labels != curr_cluster])
        b_i = np.inf
        for other_c in other_clusters:
            other_cluster_idx = np.where(labels == other_c)[0]
            avg_dist_to_other = np.mean(dist_matrix[i, other_cluster_idx])
            b_i = min(b_i, avg_dist_to_other)
            
        # 3. Считаем s(i)
        s_scores[i] = (b_i - a_i) / max(a_i, b_i)
        
    return np.mean(s_scores)

# Запуск
sil_avg = calculate_silhouette(X_pca, labels)
print(f"Итоговый силуэт модели: {sil_avg:.4f}")
```

```python
# 1. Считаем средний балл для фильтрации
raw_scores = dataset[['math score', 'reading score', 'writing score']]
mean_scores = raw_scores.mean(axis=1)

# Убираем "хвост" (например, тех, у кого средний балл < 30)
distances_from_center = np.linalg.norm(X_pca, axis=1)

# 2. Оставляем только те точки, которые входят в 95-99% распределения
# (убираем 2% самых экстремальных аутлайеров)
threshold = np.percentile(distances_from_center, 98) 
filtered_mask = distances_from_center < threshold

dataset_clean = dataset.iloc[filtered_mask].copy()
X_pca_clean = X_pca[filtered_mask]
# 2. Повторяем быстрый PCA для очищенных данных
X_clean = dataset_clean[['math score', 'reading score', 'writing score']].values
X_clean_std = (X_clean - np.mean(X_clean, axis=0)) / np.std(X_clean, axis=0)

U, S, Vt = np.linalg.svd(X_clean_std)
X_pca_clean = X_clean_std @ Vt.T[:, :2]

# 3. K-means на очищенных данных
# (Используем ваш алгоритм или сокращенную запись для скорости)
from sklearn.cluster import KMeans
kmeans_clean = KMeans(n_clusters=3, random_state=42).fit(X_pca_clean)
labels_clean = kmeans_clean.labels_

# 4. Проверка Силуэта
from sklearn.metrics import silhouette_score
new_sil = silhouette_score(X_pca_clean, labels_clean)

print(f"Старый силуэт (с хвостом): 0.4247")
print(f"Новый силуэт (без хвоста): {new_sil:.4f}")
print(f"Удалено студентов: {len(dataset) - len(dataset_clean)}")
```
